Anomaly detection is the process of identifying rare, unexpected events or data points that deviate significantly from the normal pattern of a dataset. These anomalies, often called outliers or exceptions, indicate potential errors, fraud, or malfunctions. It is essential for monitoring system health, security, and data integrity. 
* Types of Anomaly Detection
 1. Point Anomalies
A single data point is considered anomalous with respect to the rest of the data. If an individual value falls far outside the expected range, it is a point anomaly.
(a) Example: A credit card transaction of $5,000 when the typical, average spending is only $50–$100.
(b) Example: A sudden, singular spike in CPU usage to 100% in a server monitoring system.
 2. Contextual Anomalies
A data point is considered anomalous only within a specific context (e.g., time, location), even though it might appear normal in a broader, different context.
(a) Example: A temperature of 30°C (86°F) is normal in July, but it is a contextual anomaly if it occurs in January.
(b) Example: High website traffic at 3:00 AM is unusual (anomalous), whereas that same traffic level is normal at 2:00 PM.
 3. Collective Anomalies
A collection of related data points behaves unusually, even if each individual data point within that group does not appear to be an anomaly on its own.
(a) Example: A computer, over a 10-minute period, fails to log in, then sends thousands of tiny, innocuous requests. While one request is normal, the collection indicates a DDoS attack.
(b) Example: A sequence of heart rate readings that are all within a "normal" range but show a pattern that indicates arrhythmia, whereas a single reading does not.

 *  Approach and Suitable Use Cases
 1. Isolation Forest
 (a) Approach: Isolation Forest works on the principle that anomalies are few and different from normal observations, and thus are easier to "isolate" . It randomly selects features and then randomly selects a split value between the maximum and minimum values of the selected feature. This partition process creates isolation trees, and the number of splits required to isolate a data point is counted. Anomalies require fewer splits (shorter paths) on average .
 (b) Suitable Use Cases:
1. Large datasets with high dimensions: It is highly efficient and scalable for large, multi-dimensional datasets.
2. Unsupervised anomaly detection: Ideal for scenarios where no labeled data exists.
3. Network intrusion detection: Identifying unusual network traffic patterns or fraudulent transactions due to its speed and effectiveness in high-volume environments. 
 * DBSCAN (Density-Based Spatial Clustering of Applications with Noise) 
(a) Approach: DBSCAN is a clustering algorithm that defines anomalies as "noise". It identifies dense regions separated by areas of lower density. Points that fall outside these dense clusters and do not meet the minimum points required to form a cluster are classified as outliers.
(b) Suitable Use Cases:
1. Data with arbitrary shapes and sizes of clusters: Unlike algorithms that assume spherical clusters, DBSCAN can find complex cluster shapes.
2. Spatial data analysis: Excellent for identifying geographical anomalies, such as unusual geological formations or crime hotspots [1].
3. Handling noise in standard clustering tasks: It inherently separates outliers from actual clusters. 
 * Local Outlier Factor (LOF)
(a) Approach: LOF measures the local deviation of a given data point's density compared to its neighbors [1]. It defines an outlier as a point that has a significantly lower density than its immediate neighbors. A normal instance will have an LOF score near 1, while an anomaly will have a score much greater than 1.
(b) Suitable Use Cases:
1. Datasets with varying density regions: Because it considers local density, LOF can identify outliers effectively even in datasets where some areas are naturally denser than others.
2. Manufacturing defect detection: Identifying individual products that deviate from the local production norm [1].
3. Healthcare anomaly detection: Identifying patient records that show unusual combinations of vital signs relative to similar patients.

Key components of a time series include Trend (T) (long-term direction), Seasonal Variation (S) (repeating short-term cycles), Cyclical Variation (C) (long-term, non-fixed oscillations), and Irregular/Random Variation (I) (unpredictable, erratic movements). These components help break down data to analyze, forecast, and understand underlying patterns. 
 1. Trend (T): The long-term, underlying direction of the data over a long period (upward, downward, or stable).
(a) Example: An increase in global average temperatures over the past 50 years.
 2. Seasonal Variation (S): Periodic, predictable, and regular patterns that repeat within a fixed period, such as a day, week, or year.
 (a) Example: Increased sales of clothing retailers during the holiday season every December.
 3. Cyclical Variation (C): Long-term oscillations or swings around the trend, typically lasting longer than one year, often linked to economic cycles.
(a) Example: Business cycles of economic booms and recessions occurring over 5-10 years.
 4. Irregular or Random Variation (I): Unpredictable, erratic, or random fluctuations that do not follow a pattern, usually caused by unexpected, rare events.
 (a) Example: A sudden drop in stock prices following an unexpected natural disaster.

A stationary time series is one whose statistical properties, such as the mean, variance, and autocorrelation, are constant over time. This stability is a fundamental assumption for many time series modeling techniques like ARIMA.
  * Define Stationary in Time Series >
  A time series \(\{Y_{t}\}\) is considered weakly stationary (or covariance stationary, the most common practical definition) if it satisfies the following three conditions:
(a) Constant Mean: The mean value of the series is constant over time, i.e., \(E[Y_{t}]=\mu \) for all \(t\).
(b) Constant Variance: The variance of the series is constant over time, i.e., \(Var(Y_{t})=\sigma ^{2}\) for all \(t\).
(c) Constant Autocovariance: The covariance between observations at two time points depends only on the time lag (\(h\)) between them, not on the specific time points themselves, i.e., \(Cov(Y_{t},Y_{t+h})=\gamma (h)\) for all \(t\) and \(h\).
 In essence, a stationary series lacks predictable long-term trends or seasonality and tends to fluctuate around a constant long-run mean.
 * Test a Non-Stationary Series>
    You can check for stationarity using both visual inspection and formal statistical tests.
   1.  Visual Inspection >
(a) Plot the time series: Look for obvious upward or downward trends (non-constant mean) or changing volatility/amplitude (non-constant variance).
(b) Autocorrelation Function (ACF) plot: For a stationary series, the ACF will drop to zero relatively quickly. For a non-stationary series (like a random walk), the ACF tends to decay slowly and remain high for many lags.
 * Statistical Tests (Unit Root Tests)>
   (a) Augmented Dickey-Fuller (ADF) Test: This test has a null hypothesis (\(H_{0}\)) that the time series is non-stationary (it has a unit root). The series is considered stationary if the p-value is less than or equal to a significance level (e.g., 0.05), allowing rejection of the null hypothesis.
   (b) Kwiatkowski-Phillips-Schmidt-Shin (KPSS) Test: This test has a null hypothesis (\(H_{0}\)) that the time series is stationary. The series is considered non-stationary if the p-value is less than or equal to 0.05, leading to rejection of the null hypothesis.
   * Transform a Non-Stationary Series>
      Common techniques to transform a non-stationary series into a stationary one include:
1. Differencing: This involves computing the difference between consecutive observations to remove trends and stabilize the mean. The first-order difference is calculated as \(\Delta Y_{t}=Y_{t}-Y_{t-1}\). Occasionally, a second difference is needed for data with a quadratic trend.
2. Seasonal Differencing: For data with repeating seasonal patterns, you can difference the observation from the same season in the previous cycle (e.g., a lag of 12 for monthly data with yearly seasonality). This is expressed as \(\Delta _{m}Y_{t}=Y_{t}-Y_{t-m}\).
3. Logarithmic Transformation: Applying a natural logarithm (e.g., \(\log (Y_{t})\)) can help stabilize the variance when it increases over time. This transformation makes the variance more constant.
4. Detrending/Decomposition: For a series with a deterministic trend, you can explicitly model the trend (e.g., using linear regression with time as a variable) and then use the residuals for analysis.

AR, MA, ARIMA, SARIMA, and SARIMAX are time series models that progressively add complexity to forecast data: AR uses past values (\(p\)), MA uses past errors (\(q\)), ARIMA adds differencing (\(d\)) for trends, SARIMA adds seasonal components (\(P,D,Q,s\)), and SARIMAX incorporates exogenous, external variables (\(X\)). 
1. AR (Autoregressive - \(p\)): Models the current value based on a linear combination of its own previous values (\(p\) lags). Assumes stationary data.
2. MA (Moving Average - \(q\)): Models the current value based on a linear combination of previous white noise error terms (\(q\) lags).
3. ARIMA (Autoregressive Integrated Moving Average - \(p,d,q\)): Combines AR and MA with differencing (\(d\)) to make non-stationary data stationary (remove trends).
4. SARIMA (Seasonal ARIMA - \(p,d,q\times P,D,Q,s\)): Extends ARIMA to handle seasonality by adding seasonal autoregressive (\(P\)), seasonal differencing (\(D\)), and seasonal moving average (\(Q\)) components over a period (\(s\)).
5. SARIMAX (SARIMA with eXogenous variables - \(p,d,q\times P,D,Q,s+X\)): A SARIMA model that includes external, independent variables (\(X\)) to improve forecasting accuracy (e.g., promotions, holidays).
   * Key Differences and Applications >
   1. AR/MA: Basic stationary, non-seasonal data forecasting.
   2. ARIMA: Data with trends (non-stationary) but no seasonality (e.g., stock price trends).
   3. SARIMA: Data with both trends and strong seasonality (e.g., monthly sales, weather patterns).
   4. SARIMAX: Complex data with seasonality, trend, and external influences (e.g., forecasting sales based on marketing spend).

Question 6: Load a time series dataset (e.g., AirPassengers), plot the original series, and decompose it into trend, seasonality, and residual components (Include your Python code and output in the code box below.) 

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Load the AirPassengers dataset
# Dataset is available online as a CSV
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
df = pd.read_csv(url)

# Convert 'Month' to datetime and set as index
df['Month'] = pd.to_datetime(df['Month'])
df.set_index('Month', inplace=True)

# Rename column for clarity
df.columns = ['Passengers']

# Plot the original time series
plt.figure(figsize=(10, 4))
plt.plot(df['Passengers'])
plt.title("AirPassengers Time Series")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.show()

# Decompose the time series (multiplicative due to increasing variance)
decomposition = seasonal_decompose(df['Passengers'], model='multiplicative', period=12)

# Plot decomposition components
decomposition.plot()
plt.show()
```

Question 7: Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to detect anomalies. Visualize the anomalies on a 2D scatter plot. (Include your Python code and output in the code box below.) 

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# -------------------------------
# Create NYC Taxi Fare-like Dataset
# -------------------------------
np.random.seed(42)

# Normal taxi trips
fare_amount = np.random.normal(loc=15, scale=5, size=500)
trip_distance = np.random.normal(loc=3, scale=1, size=500)

# Anomalous trips (very high fare or distance)
fare_anomaly = np.random.uniform(60, 100, size=25)
distance_anomaly = np.random.uniform(15, 25, size=25)

# Combine data
fare = np.concatenate([fare_amount, fare_anomaly])
distance = np.concatenate([trip_distance, distance_anomaly])

df = pd.DataFrame({
    'fare_amount': fare,
    'trip_distance': distance
})

# -------------------------------
# Apply Isolation Forest
# -------------------------------
iso_forest = IsolationForest(
    n_estimators=100,
    contamination=0.05,
    random_state=42
)

df['anomaly'] = iso_forest.fit_predict(df[['fare_amount', 'trip_distance']])

# Separate normal and anomalous points
normal = df[df['anomaly'] == 1]
anomalies = df[df['anomaly'] == -1]

# -------------------------------
# Visualization
# -------------------------------
plt.figure(figsize=(8, 5))
plt.scatter(normal['fare_amount'], normal['trip_distance'], alpha=0.6, label='Normal')
plt.scatter(anomalies['fare_amount'], anomalies['trip_distance'],
            color='red', label='Anomaly')
plt.xlabel("Fare Amount")
plt.ylabel("Trip Distance")
plt.title("Isolation Forest Anomaly Detection (NYC Taxi Fare)")
plt.legend()
plt.show()
```

Question 8: Train a SARIMA model on the monthly airline passengers dataset. Forecast the next 12 months and visualize the results. (Include your Python code and output in the code box below.) 

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX

# -------------------------------
# Load AirPassengers Dataset (Manually)
# -------------------------------
data = [
    112,118,132,129,121,135,148,148,136,119,104,118,
    115,126,141,135,125,149,170,170,158,133,114,140,
    145,150,178,163,172,178,199,199,184,162,146,166,
    171,180,193,181,183,218,230,242,209,191,172,194,
    196,196,236,235,229,243,264,272,237,211,180,201,
    204,188,235,227,234,264,302,293,259,229,203,229,
    242,233,267,269,270,315,364,347,312,274,237,278,
    284,277,317,313,318,374,413,405,355,306,271,306,
    315,301,356,348,355,422,465,467,404,347,305,336,
    340,318,362,348,363,435,491,505,404,359,310,337,
    360,342,406,396,420,472,548,559,463,407,362,405,
    417,391,419,461,472,535,622,606,508,461,390,432
]

dates = pd.date_range(start="1949-01", periods=len(data), freq="M")
df = pd.DataFrame(data, index=dates, columns=["Passengers"])

# -------------------------------
# Train SARIMA Model
# -------------------------------
model = SARIMAX(
    df["Passengers"],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

results = model.fit(disp=False)

# -------------------------------
# Forecast Next 12 Months
# -------------------------------
forecast = results.get_forecast(steps=12)
forecast_mean = forecast.predicted_mean
conf_int = forecast.conf_int()

# -------------------------------
# Visualization
# -------------------------------
plt.figure(figsize=(10,5))
plt.plot(df["Passengers"], label="Observed")
plt.plot(forecast_mean, label="Forecast", color="red")
plt.fill_between(
    conf_int.index,
    conf_int.iloc[:, 0],
    conf_int.iloc[:, 1],
    color="pink",
    alpha=0.3
)
plt.title("SARIMA Forecast – AirPassengers (Next 12 Months)")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.legend()
plt.show()
```

Question 9: Apply Local Outlier Factor (LOF) on any numerical dataset to detect anomalies and visualize them using matplotlib. (Include your Python code and output in the code box below.) 

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor

# -------------------------------
# Create a numerical dataset
# -------------------------------
np.random.seed(42)

# Normal data points
x_normal = np.random.normal(loc=10, scale=2, size=300)
y_normal = np.random.normal(loc=10, scale=2, size=300)

# Anomalous data points
x_outliers = np.random.uniform(20, 30, size=20)
y_outliers = np.random.uniform(20, 30, size=20)

# Combine data
x = np.concatenate([x_normal, x_outliers])
y = np.concatenate([y_normal, y_outliers])

df = pd.DataFrame({'X': x, 'Y': y})

# -------------------------------
# Apply Local Outlier Factor
# -------------------------------
lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.06
)

df['anomaly'] = lof.fit_predict(df[['X', 'Y']])

# Separate normal points and anomalies
normal = df[df['anomaly'] == 1]
anomalies = df[df['anomaly'] == -1]

# -------------------------------
# Visualization
# -------------------------------
plt.figure(figsize=(8, 5))
plt.scatter(normal['X'], normal['Y'], alpha=0.6, label='Normal')
plt.scatter(anomalies['X'], anomalies['Y'], color='red', label='Anomaly')
plt.xlabel("Feature X")
plt.ylabel("Feature Y")
plt.title("Local Outlier Factor (LOF) Anomaly Detection")
plt.legend()
plt.show()
```

Question 10: You are working as a data scientist for a power grid monitoring company. Your goal is to forecast energy demand and also detect abnormal spikes or drops in real-time consumption data collected every 15 minutes. The dataset includes features like timestamp, region, weather conditions, and energy usage. 
Explain your real-time data science workflow: 
● How would you detect anomalies in this streaming data (Isolation Forest / LOF / DBSCAN)?
● Which time series model would you use for short-term forecasting (ARIMA / SARIMA / SARIMAX)? 
● How would you validate and monitor the performance over time? 
● How would this solution help business decisions or operations? 
(Include your Python code and output in the code box below.) 

```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from statsmodels.tsa.statespace.sarimax import SARIMAX

# -------------------------------
# Simulated 15-minute energy data
# -------------------------------
np.random.seed(42)

date_range = pd.date_range(start="2024-01-01", periods=500, freq="15min")
energy_usage = np.random.normal(100, 10, size=500)

# Inject anomalies
energy_usage[100] = 180
energy_usage[300] = 40

temperature = np.random.normal(25, 5, size=500)

df = pd.DataFrame({
    "timestamp": date_range,
    "energy_usage": energy_usage,
    "temperature": temperature
}).set_index("timestamp")

# -------------------------------
# Anomaly Detection (Isolation Forest)
# -------------------------------
iso = IsolationForest(contamination=0.02, random_state=42)
df["anomaly"] = iso.fit_predict(df[["energy_usage"]])

# -------------------------------
# SARIMAX Forecasting
# -------------------------------
model = SARIMAX(
    df["energy_usage"],
    exog=df[["temperature"]],
    order=(1,1,1),
    seasonal_order=(1,1,1,96)  # daily seasonality (96 x 15min)
)

results = model.fit(disp=False)

forecast = results.get_forecast(
    steps=20,
    exog=df[["temperature"]].iloc[-20:]
)

# -------------------------------
# Visualization
# -------------------------------
plt.figure(figsize=(10,5))
plt.plot(df["energy_usage"], label="Observed")
plt.scatter(
    df[df["anomaly"] == -1].index,
    df[df["anomaly"] == -1]["energy_usage"],
    color="red",
    label="Anomaly"
)
plt.plot(forecast.predicted_mean, color="green", label="Forecast")
plt.title("Energy Demand Forecasting & Anomaly Detection")
plt.legend()
plt.show()
```
