Question 2: Explain overfitting and underfitting in ML. How can you detect and prevent them? Hint: Discuss bias-variance tradeoff, cross-validation, and regularization techniques

 This is The Answer>>
  Overfitting and underfitting are common challenges in machine learning that impact a model's ability to generalize to unseen data.
 1. Overfitting:
Overfitting occurs when a model learns the training data too well, including noise and outliers, leading to excellent performance on the training set but poor performance on new, unseen data. This indicates high variance and low bias.
 2. Underfitting:
Underfitting occurs when a model is too simple to capture the underlying patterns in the training data, resulting in poor performance on both the training and test sets. This indicates high bias and low variance. 
  * Bias-Variance Tradeoff:
Overfitting and underfitting are two sides of the bias-variance tradeoff. 
 1.Bias refers to the error introduced by approximating a real-world problem, which may be complex, with a simplified model. High bias leads to underfitting.
 2. Variance refers to the model's sensitivity to small fluctuations in the training data. High variance leads to overfitting.
The goal is to find a balance where the model is complex enough to capture the data's patterns (low bias) but not so complex that it memorizes noise (low variance).
  * Detection:
  Performance on Training vs. Test Data:
  1.  Overfitting: High training accuracy/low training error, but significantly lower test accuracy/higher test error.
  2. derfitting: Low training accuracy/high training error, and similarly low test accuracy/high test error.
 3. Learning Curves: Plotting training and validation loss/accuracy over epochs can reveal divergence (overfitting) or consistently high error (underfitting).
  
  * Prevention:
  1. Cross-Validation: Techniques like k-fold cross-validation help assess model performance on different subsets of the data, providing a more robust estimate of generalization error and helping detect overfitting.
  * Regularization Techniques:
  (a) L1 (Lasso) and L2 (Ridge) Regularization: Add a penalty to the loss function based on the magnitude of model coefficients, discouraging overly complex models and shrinking coefficients towards zero.
  (b) Dropout (for Neural Networks): Randomly deactivates a percentage of neurons during training, preventing co-adaptation and forcing the network to learn more robust features.
  1. Data Augmentation: Increases the size and diversity of the training data, especially useful for image data, to help the model generalize better.
  2. Early Stopping: In iterative training algorithms, stop training when the performance on a validation set starts to degrade, preventing the model from overfitting to the training data.
  3. Feature Selection/Engineering: Selecting relevant features and creating new, more informative features can simplify the model and reduce the risk of overfitting.
 4. Reduce Model Complexity: For complex models like deep neural networks, reducing the number of layers or neurons can help prevent overfitting.

 Question 3:How would you handle missing values in a dataset? Explain at least three methods with examples. Hint: Consider deletion, mean/median imputation, and predictive modeling. 

This is The Answer>>
 You can handle missing values by deleting rows or columns, imputing with a central tendency like the mean or median, or using more advanced predictive modeling to estimate the missing values. The best method depends on the amount of missing data, the nature of the data (e.g., time-series), and the goals of the analysis. 
 1. Deletion
 (a) What it is: Removing rows (listwise deletion) or columns that contain missing values.
 (b) Example: If you have a dataset of 100 customers and 5 of them are missing an "age" value, you could delete those 5 rows to analyze the remaining 95. If a feature like "phone number" has many missing values, you might delete that entire column.
 (c) Best for: Small amounts of missing data where the missing values are not systematically related to other variables.
 (d) Caution: This can lead to a significant loss of information and introduce bias if the missing data is not random. 
 2. Mean/Median Imputation
 (a) What it is: Replacing missing values with the mean (average) or median (middle value) of the non-missing values in that column.
  (b) Example: In a column of "income," if there are some missing values, you could calculate the average income of all the people with income data and replace the missing values with that average. For a column with a skewed distribution, using the median is often better to avoid skewing the data.
 (c) Best for: Numerical data with a small number of missing values, when the data is not time-series based.
 (d) Caution: This can reduce the variance of the dataset and distort the relationships between variables if the missingness is not random. 
 3. Predictive Modeling
 (a) What it is: Using a predictive model to estimate and fill in the missing values based on other features in the dataset.
 (b) Example: You could build a regression model to predict a missing "house price" based on other features like "square footage" and "number of bedrooms". The model learns the relationship from the complete data points and then uses it to predict the prices for the missing values.
 (c) Best for: When there are relationships between the variables in the dataset and a more accurate imputation is needed.
 (d) Caution: This is a more complex method that assumes the relationships in the complete data hold true for the incomplete data. The accuracy of the imputed values depends on the strength of the predictive model. 


Question 4:What is an imbalanced dataset? Describe two techniques to handle it (theoretical + practical). Hint: Discuss SMOTE, Random Under/Oversampling, and class weights in models. 

This is the answer>>
An imbalanced dataset is a dataset in which the distribution of class labels is highly unequal.
For example, in a binary classification problem (fraud vs. non-fraud), one class may make up 95–99% of the samples, while the minority class accounts for only 1–5%.

This imbalance can cause machine-learning models to become biased toward the majority class, leading to poor detection of the minority class (often the class of interest).

Two techniques to handle imbalanced datasets
1. SMOTE (Synthetic Minority Oversampling Technique)

 * Theoretical explanation:
SMOTE creates synthetic samples of the minority class rather than duplicating existing ones.
It takes a minority-class sample, finds its k nearest minority neighbors, and generates new samples by interpolating between them.
This helps the model learn better decision boundaries without simple repetition.
  * Practical example:

from imblearn.over_sampling import SMOTE

smote = SMOTE(k_neighbors=5)
X_resampled, y_resampled = smote.fit_resample(X, y)


Use when you have enough continuous features and want to avoid overfitting that may come from naïve oversampling.

  2. Random Undersampling or Oversampling

 * Theoretical explanation:

 1. Random Oversampling: Duplicate minority-class samples until classes are balanced.

 2. Random Undersampling: Remove random samples from the majority class.

Oversampling increases dataset size (risk of overfitting), whereas undersampling reduces training time but risks losing useful information.

 * Practical example:

from imblearn.over_sampling import RandomOverSampler
from imblearn.under_sampling import RandomUnderSampler

ros = RandomOverSampler()
X_over, y_over = ros.fit_resample(X, y)

rus = RandomUnderSampler()
X_under, y_under = rus.fit_resample(X, y)

  * Alternative technique: Class Weights (in models)

(Useful to mention for completeness)

  * Theoretical explanation:
Instead of changing the dataset, adjust the loss function so mistakes on the minority class are penalized more heavily.
This forces the model to pay more attention to the minority examples.

 * Practical example (e.g., Logistic Regression):

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(class_weight='balanced')
model.fit(X, y)
 * Summary > 
  Technique                 Type              Advantage                              Disadvantage                       
| ---------------------| ------------    | --------------------------------    | ------------------------------------- |
|  SMOTE               | Oversampling    | Creates realistic synthetic samples | May create noisy points               |
| Random Oversampling  | Oversampling    | Simple and effective                | Can overfit                           |
| Random Undersampling | Undersampling   | Fast and reduces size               | Risk of losing information            |
| Class Weights        | Algorithm-level | No data manipulation                | May not fully solve extreme imbalance |


Question 5: Why is feature scaling important in ML? Compare Min-Max scaling and Standardization. Hint: Explain impact on distance-based algorithms (e.g., KNN, SVM) and gradient descent. 

 This is The Answer >> 
 Feature scaling is a crucial preprocessing step in machine learning that transforms features of varying scales into a comparable range. This is important because many machine learning algorithms are sensitive to the magnitude of features. 
 * Impact on Distance-Based Algorithms (e.g., KNN, SVM):>>
 Distance-based algorithms like K-Nearest Neighbors (KNN) and Support Vector Machines (SVM) rely on calculating distances between data points. If features have drastically different scales (e.g., age in years vs. income in thousands), features with larger magnitudes will disproportionately influence the distance calculations. This can lead to biased results where the algorithm prioritizes features with larger numerical ranges, even if they are not inherently more important. Scaling ensures that all features contribute equally to the distance metric, preventing one feature from dominating the decision-making process.
 * Impact on Gradient Descent:
  Algorithms that use gradient descent for optimization, such as linear regression, logistic regression, and neural networks, also benefit significantly from feature scaling. Without scaling, features with larger ranges will have larger gradients, leading to larger steps in the optimization process along those dimensions. This can cause the optimization path to oscillate inefficiently and slow down convergence, making it harder to reach the optimal solution. Scaling creates a more spherical or well-conditioned loss surface, allowing gradient descent to converge faster and more smoothly.
  * Comparison of Min-Max Scaling and Standardization:>
   1. Min-Max Scaling (Normalization):
      Formula:
      x<sub>scaled</sub> = x - x<sub>min</sub> / x<sub>max</sub> - x<sub>min</sub>
                
  3. Range: Transforms features to a specific range, typically between 0 and 1.
  4. Sensitivity to Outliers: Highly sensitive to outliers, as they directly influence the minimum and maximum values, potentially compressing the range of other data points.
 5. Use Cases: Suitable when the data does not follow a normal distribution and for algorithms that do not assume any specific data distribution, like KNN and neural networks, especially when a bounded output range is desired.
 *  Standardization (Z-score Normalization):
  1. Formula:
     x<sub>scaled</sub> = x - μ / σ

   2. Range: Transforms features to have a mean of 0 and a standard deviation of 1. The values are not bounded to a specific range.
   3. Sensitivity to Outliers: Less sensitive to outliers compared to Min-Max scaling, as it uses the mean and standard deviation, which are more robust to extreme values.
  4. Use Cases: Preferred when the data follows a normal distribution or for algorithms that assume a Gaussian distribution (e.g., Linear Regression, Logistic Regression, Support Vector Machines). It is also generally a good choice when outliers are present.

Question 6:  Compare Label Encoding and One-Hot Encoding. When would you prefer one over the other? Hint: Consider categorical variables with ordinal vs. nominal relationships. 

This is The Answer>>
 You should use Label Encoding for ordinal data (where categories have a natural order) and One-Hot Encoding for nominal data (where categories have no inherent order). Label Encoding assigns a unique integer to each category, which is efficient but can incorrectly imply an order to nominal data, while One-Hot Encoding creates a new binary column for each category, avoiding the false ordering issue but increasing dimensionality. 
 Feature>>
  * label coding>>
 1. data type>Ordinal (e.g., low, medium, high)
 2. Output>	A single integer for each category
 3. Pros>	Simple, memory-efficient, and suitable for tree-based models
 4. Cons>	Can impose an artificial order on nominal data, leading to misleading relationships
    
  * one hot encoding>
     1. Data Type> Nominal (e.g., color, city)
     2. output> Multiple binary (0 or 1) columns, one for each category
     3. pros> Prevents false ordinality, works well for linear models and neural networks
     4. cons> Increases dimensionality, which can lead to high memory usage and computational complexity (the "curse of dimensionality")
        
       * When to prefer one over the other
  1. Prefer Label Encoding when:
 (a) Your categorical data has a clear, natural order, such as education levels (High School, Bachelor's, Master's) or ratings (low, medium, high).
 (b) You are using algorithms that can handle this type of encoded data well, like decision trees or random forests.
 (c) Memory and computational efficiency are critical, especially with a very large number of categories.
 2. Prefer One-Hot Encoding when:
 (a) Your categorical data has no natural order, such as a list of colors or country names.
 (b) You want to avoid any potential bias from a false numerical ordering.
 (c) You are using algorithms that assume no ordinality in the data, like linear regression or neural networks.
 (d) You have sufficient memory and the number of unique categories is manageable, as it avoids the "curse of dimensionality".

```python
import pandas as pd
from io import StringIO

df = pd.read_csv("titanic.csv")

# --- Part (a): Survival rate by Pclass ---
survival_by_class = df.groupby("Pclass")["Survived"].mean()

# --- Part (b): Children vs Adults ---
df["AgeGroup"] = df["Age"].apply(lambda x: "Child" if pd.notnull(x) and x < 18 else "Adult")
survival_by_agegroup = df.groupby("AgeGroup")["Survived"].mean()

print("Survival Rate by Pclass:")
print(survival_by_class)
print("\nSurvival Rate by Age Group:")
print(survival_by_agegroup)
```

Question 10:  HR Analytics Dataset a). What factors most strongly correlate with employee attrition? Use visualizations to show key drivers (e.g., satisfaction, overtime, salary). b). Are employees with more projects more likely to leave?  Dataset: hr_analytics 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV
df = pd.read_csv('hr_analytics.csv')

# a) Factors most strongly correlated with attrition

salary_order = sorted(df['salary'].unique())
salary_mapping = {level: i for i, level in enumerate(salary_order)}
df['salary_num'] = df['salary'].map(salary_mapping)
df['sales_num'] = df['sales'].astype('category').cat.codes

numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
correlations = df[numeric_cols].corr()['left'].sort_values(ascending=False)
print("\nCorrelation with attrition (left):")
print(correlations)

plt.figure(figsize=(10,6))
sns.barplot(x=correlations.index, y=correlations.values, palette='coolwarm')
plt.title("Correlation of Numeric Features with Employee Attrition")
plt.xticks(rotation=45)
plt.show()

# Visualize key factors vs attrition
key_factors = ['satisfaction_level', 'Work_accident', 'salary_num', 'time_spend_company']
for factor in key_factors:
    plt.figure(figsize=(8,4))
    sns.boxplot(x='left', y=factor, data=df)
    plt.title(f"{factor} vs Attrition")
    plt.show()


# b) Are employees with more projects more likely to leave? - NO
plt.figure(figsize=(8,4))
sns.boxplot(x='left', y='number_project', data=df)
plt.title("Number of Projects vs Employee Attrition")
plt.show()

mean_projects = df.groupby('left')['number_project'].mean()
print("\nAverage number of projects by attrition:")
print(mean_projects)
```
