   * Abstraction and encapsulation are both fundamental principles in object-oriented programming (OOP), but they address different aspects of software design:
   * Abstraction:
  1. Focus: Hiding complex implementation details and showing only the essential features or functionalities to the user. It focuses on what an object does, not how it does it.
  2. Purpose: To simplify complex systems by providing a clear and concise interface, making the code easier to understand, use, and maintain.

You don't need to understand the intricate workings of the engine or transmission (the hidden implementation details) to drive it.   * Encapsulation:
  1. Focus: Bundling data (attributes) and methods (functions) that operate on that data into a single unit, typically a class, and controlling access to that data. It focuses on how an object's data is accessed and manipulated.
  2. Purpose: To protect data from unauthorized access or modification, ensuring data integrity and consistency. It promotes modularity and makes code more robust.
  Example: In a Car class, the engine's internal state (e.g., fuel level, RPM) might be encapsulated within the class and only accessible through specific methods like start_engine() or get_fuel_level(), preventing direct manipulation that could lead to errors. In Python, encapsulation is achieved by using conventions for private attributes (e.g., _private_attribute) and providing public methods to interact with them. 
  * Key Differences Summarized:
  1. Goal: Abstraction aims to simplify by hiding complexity; Encapsulation aims to protect data by bundling it with its operations and controlling access.
  2. What is hidden: Abstraction hides implementation details; Encapsulation hides the internal state and structure of an object.
  3. Perspective: Abstraction is user-focused (what the user sees and interacts with); Encapsulation is developer-focused (how the object is structured and protected).

  * Polymorphism in Object-Oriented Programming (OOP) is the ability of objects and functions to take on multiple forms or behaviors through a single interface or reference. This key OOP concept allows a single method call to invoke different implementations depending on the object's type, leading to more flexible, extensible, and reusable code. It is achieved through techniques like method overloading (compile-time polymorphism) and method overriding (runtime polymorphism), which enables objects of different classes to be treated as instances of a common superclass.  
  * How it Works
  1. A Common Interface: Polymorphism provides a single interface to represent various data types.
  2. Different Forms: Objects of different classes can be treated as objects of a common superclass or interface, allowing them to be handled uniformly.

  2. Runtime Polymorphism (Method Overriding): This involves a function in a derived (child) class having the same name, return type, and parameters as a function in its base (parent) class. The specific function to be executed is resolved at runtime, allowing the child class to provide its own unique implementation. 
  * Benefits of Polymorphism
  1. Flexibility: Programs can handle objects of different types in a uniform way, making them adaptable to new requirements. 
  2. Extensibility: New classes can be easily created from existing ones, extending the program's functionality without modifying the original code. 
  3. Code Reusability: Polymorphism reduces code redundancy and the need to write similar code repeatedly, promoting modularity.

Unlike some other object-oriented programming languages like Java or C++, Python does not support "true" method overloading in the sense of defining multiple methods with identical names but distinct signatures.
Instead, Python achieves a similar effect and provides flexibility in handling different argument scenarios within a single method definition through mechanisms such as:
 1. Default Arguments: Assigning default values to parameters allows a method to be called with or without providing values for those parameters.

        def greet(self, name="Guest"):
            print(f"Hello, {name}!")

    g = Greeter()
    g.greet()         # Output: Hello, Guest!
    g.greet("Alice")  # Output: Hello, Alice!
 .Variable-Length Arguments (*args and `: kwargs`):**
 (a) *args allows a method to accept an arbitrary number of positional arguments as a tuple.
 (b) **kwargs allows a method to accept an arbitrary number of keyword arguments as a dictionary.
Python

    class Calculator:
        def add(self, *numbers):
            return sum(numbers)

        def display_info(self, **details):
            for key, value in details.items():
                print(f"{key}: {value}")

    calc = Calculator()
    print(calc.add(1, 2, 3))  # Output: 6
    calc.display_info(name="Bob", age=30) # Output: name: Bob, age: 30
  * Conditional Logic within a Single Method: Implementing if/else statements or other conditional checks within a method to determine behavior based on the types or number of arguments received.
Python

    class Processor:
        def process(self, data):
            if isinstance(data, int):
                print(f"Processing integer: {data * 2}")
            elif isinstance(data, str):
                print(f"Processing string: {data.upper()}")

                print("Cannot process this data type.")

    p = Processor()
    p.process(10)      # Output: Processing integer: 20
    p.process("hello") # Output: Processing string: HELLO
These techniques enable Python developers to create versatile methods that can adapt to various input scenarios, effectively achieving the benefits of method overloading without requiring multiple distinct method definitions with the same name.

  This is the answer>>
  * Method overriding in OOP is when a subclass provides its own specific implementation of a method that already exists in its parent class, using the exact same method name, return type, and parameters. This technique allows for runtime polymorphism (also known as dynamic or late binding), where the correct method is determined at runtime based on the actual object type, and it lets a child class customize inherited behaviors without changing the parent class. 
  * Key Characteristics
  1. Inheritance is required: Method overriding only occurs between a parent class (superclass) and a child class (subclass) through inheritance. 
  2. Same method signature: The overridden method in the child class must have the same name, return type, and parameter list as the method in the parent class. 
  3. Runtime decision (Dynamic Binding): The specific method that gets executed is decided at runtime based on the actual object's type, not the reference variable's type. 
  4. Customized behavior: It allows the subclass to provide its unique implementation for an inherited method, tailoring the behavior to its specific needs. 
  5. Polymorphism: Method overriding is a core concept of runtime polymorphism, enabling a single interface to be used for a general class of actions.

This is the answer>>
  * The difference between a class variable and an instance variable lies in their scope, ownership, and how they are accessed and modified.
  * Class Variable:
  1. Scope and Ownership: A class variable belongs to the class itself, not to any specific instance of the class. There is only one copy of a class variable, which is shared by all instances of that class.
  2. Creation and Lifetime: Class variables are created when the class is loaded into memory and exist for the entire duration the class is in memory (typically until the program ends).
  3. Access: They can be accessed using the class name (e.g., ClassName.variable_name) or through an instance of the class (e.g., object_name.variable_name).
 4. Use Cases: Class variables are suitable for storing data that is common to all instances of a class, such as constants, shared configurations, or counters for the number of instances created. 
  * Instance Variable:
 1. Scope and Ownership: An instance variable belongs to a specific instance (object) of a class. Each object has its own independent copy of the instance variables. 
 2. Creation and Lifetime: Instance variables are created when an object is instantiated (e.g., using the new keyword in Java or by calling the class constructor in Python) and are destroyed when the object is no longer referenced and garbage collected. 
 3. Access: They are accessed using an object reference (e.g., object_name.variable_name).
 4. Use Cases: Instance variables are used to store unique data or characteristics specific to each individual object, such as the name, age, or ID of a person object.

This is the answer>>
  * The super() function in Python holds significant importance within the context of object-oriented programming, particularly concerning inheritance. Its primary role is to provide a mechanism for calling methods and accessing attributes of a parent or sibling class from a child or subclass.
* Here's a breakdown of its key significances:
 1. Method Overriding and Extension: When a child class overrides a method inherited from its parent, super() allows the child to still call the parent's version of that method. This enables the child class to extend or augment the parent's functionality rather than completely replacing it. For instance, in __init__ methods, super().__init__() ensures the parent's initialization logic is executed before the child's specific initialization.
 2. Code Reusability and Maintainability: By using super(), you avoid explicitly naming the parent class when calling its methods. This promotes code reusability as it makes your code more adaptable to changes in the inheritance hierarchy. If a parent class name changes, you don't need to update every child class that calls its methods; super() handles this dynamically.
 3. Handling Multiple Inheritance (MRO): In scenarios with multiple inheritance, super() intelligently navigates the Method Resolution Order (MRO) to determine which parent or sibling method to call. This ensures that the correct method in the inheritance chain is invoked, adhering to the order defined by the MRO.
 4. Forward Compatibility: super() enhances forward compatibility. If the class hierarchy changes in the future, code using super() is more likely to remain functional with minimal adjustments, as it relies on the MRO rather than hardcoded parent class names.
In essence, super() is a cornerstone of robust and flexible object-oriented design in Python, facilitating cleaner code, better maintainability, and effective management of inheritance relationships, especially in complex scenarios like multiple inheritance.

  * Important Considerations:
 1. Not Guaranteed Timing: The exact timing of __del__ execution is not guaranteed. Python's garbage collector determines when an object is actually destroyed, which can be delayed, especially in cases of circular references.
 2. Avoid for Critical Cleanup: Due to the non-guaranteed timing, __del__ should not be relied upon for critical cleanup tasks where immediate resource release is essential. For such scenarios, context managers (with statements) are the preferred and more reliable approach.
 3. Error Handling: Exceptions raised within __del__ are typically ignored, which can lead to silent failures and make debugging challenging.
In essence, __del__ offers a way to manage external resources and perform finalization steps, but its use requires careful consideration of its limitations and the availability of more robust alternatives like context managers for critical resource management.

This is the answer>>
  * Method chaining in Python Object-Oriented Programming (OOP) is a technique where multiple method calls are invoked sequentially on the same object within a single expression. This is achieved by designing methods within a class such that each method, after performing its operation, returns the object itself (usually self).
  * How it works:
 1. Return self: For method chaining to work, each method intended to be part of a chain must explicitly return self at the end of its execution. This ensures that the method call effectively returns the instance of the object it was called on, allowing the next method in the chain to operate on the same object.
 2. Sequential Invocation: When a method returns self, the result of that method call is the object itself. This allows you to immediately append another method call to the returned object, creating a "chain" of operations.
 * Benefits of Method Chaining:
 1. Readability: It can make code more concise and readable, especially when performing a series of related operations on an object, as it reduces the need for intermediate variables.
 2. Conciseness: It eliminates the need to repeatedly reference the object name for each method call.
 3. Fluent APIs: It facilitates the creation of "fluent APIs" where code reads more like natural language, enhancing expressiveness.

```python
from abc import ABC, abstractmethod
import math

# Abstract base class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

# Derived class for Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

# Derived class for Rectangle
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Example usage
circle = Circle(5)
rectangle = Rectangle(4, 6)

print(f"Circle area: {circle.area():.2f}")       
print(f"Rectangle area: {rectangle.area()}")  
```

```python
Q5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes 
balance and methods to deposit, withdraw, and check balance.
```

```python
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance  # Private attribute

    # Method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ${amount}")
        else:
            print("Invalid deposit amount.")

    # Method to withdraw money
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn: ${amount}")
        else:
            print("Insufficient balance or invalid amount.")

    # Method to check balance
    def get_balance(self):
        print(f"Current Balance: ${self.__balance}")
        return self.__balance

# Example usage
account = BankAccount(100)  
account.get_balance()       

account.deposit(50)       
account.get_balance()       

account.withdraw(30)        
account.get_balance()    

account.withdraw(200)      
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overload the + operator
    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented

    # String representation
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# Example usage
v1 = Vector(2, 3)
v2 = Vector(4, 5)

v3 = v1 + v2  # Uses overloaded __add__ method

print(v1) 
print(v2) 
print(v3)  
```

```python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades  # Expecting a list of numbers

    def average_grade(self):
        if not self.grades:
            return 0  # Avoid division by zero if no grades
        return sum(self.grades) / len(self.grades)

# Example usage
student = Student("John", [85, 90, 78, 92])
print(f"{student.name}'s average grade is: {student.average_grade():.2f}")
```

```python
Q.14 Create a class Employee with a method calculate_salary() that computes the salary based on hours worked 
and hourly rate. Create a derived class Manager that adds a bonus to the salary.
```

```python
# Base class
class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

# Derived class
class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        super().__init__(name, hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        base_salary = super().calculate_salary()
        return base_salary + self.bonus

# Example usage
emp = Employee("ram", 40, 20)
print(f"{emp.name}'s salary: ${emp.calculate_salary()}")  

mgr = Manager("shyam", 40, 30, 500)
print(f"{mgr.name}'s salary: ${mgr.calculate_salary()}")  
```
